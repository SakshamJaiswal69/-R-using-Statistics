```r
# Create two variables a = 20 and b = 10. Find:

# Addition
# Subtraction
# Multiplication
# Division
# Remainder
# Power
a = 25
b=65
print(a+b) #  + -> Addition
print(b-a) #  - -> Subtraction
print(a*b) #  * -> Multiplication
print(a/b) #  / -> Division
print(a%%b) # %% -> Modulo(remainder)
print(a%/%b) # %/% -> Integer division
print(b**a) # ^ or **-> Exponention
```
